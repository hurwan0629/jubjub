```python
from pathlib import Path
TACO_DIR = Path("../datasets/TACO/data/taco_yolo").resolve()

TACO_DIR.exists()
```

```python
import json
from collections import Counter

meta = json.loads((TACO_DIR.parent / "annotations.json").read_text(encoding="utf-8"))
```

```python
counts = {}

for split in ("train", "val"):
  counts[split] = Counter()

  for label_path in (TACO_DIR / "labels" / split).glob("*.txt"):
    for line in label_path.read_text(encoding="utf-8").splitlines():
      if line.strip():
        class_id = int(line.split()[0])
        counts[split][class_id] += 1

print(f"{"클래스":29} {"train":>5} {"val":>6} {"val 비율":>9}")
for category in meta["categories"]:
  class_id = category["id"]
  train = counts["train"][class_id]
  val = counts["val"][class_id]
  ratio = val / (train + val) if train + val else 0
  print(f"{category["name"][:30]:30} {train:>6} {val:>6} {ratio:>10.1%} {category["name"] + " 없음" if val == 0 else "":>25}")
```

```python
sup_counts = {}
printed = set()

# split별로 sup_conuts 추가해주기.
for split in ("train", "val"):
  sup_counts[split] = Counter()

  # 모든 supercategory에 대해 순회
  for super_cat in list(map(lambda x: x["supercategory"], meta["categories"])):
    # if split == "train" and super_cat not in printed:
    #   printed.add(super_cat)
      
    # train과 val에 대해서 supercategory: 개수 로 만들어주기
    sup_counts[split][super_cat] = sum([counts[split][c["id"]] for c in list(filter(lambda x: x["supercategory"] == super_cat, meta["categories"]))])

for super_cat in list(map(lambda x: x["supercategory"], meta["categories"])):
    if super_cat not in printed:
      printed.add(super_cat)
      print(f"{super_cat}: {list(map(lambda x: x["name"], filter(lambda x: x["supercategory"] == super_cat, meta["categories"])))}")
      print(sup_counts["train"][super_cat], sup_counts["val"][super_cat])
```

```python
import matplotlib.pyplot as plt

splits = ("train", "val")
supercategories = sorted(
    {c["supercategory"] for c in meta["categories"]}
)
bottom = [0, 0]

plt.figure(figsize=(9, 7))

for supercategory in supercategories:
    heights = [
        sup_counts[split][supercategory] / sum(sup_counts[split].values()) * 100
        for split in splits
    ]
    plt.bar(splits, heights, bottom=bottom, label=supercategory)
    bottom = [bottom[i] + heights[i] for i in range(2)]

plt.ylabel("object ratio (%)")
plt.ylim(0, 100)
plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
from ultralytics import YOLO

weight_path = Path(
    "train/weights/best.pt"
)
trained_model = YOLO(weight_path)

image_paths = sorted((TACO_DIR / "images" / "val").glob("*"))
image_path = image_paths[0]  # 다른 이미지를 보려면 인덱스 변경
label_path = TACO_DIR / "labels" / "val" / f"{image_path.stem}.txt"

image = Image.open(image_path).convert("RGB")
width, height = image.size
prediction = trained_model.predict(image, conf=0.10, verbose=False)[0]

fig, (gt_ax, pred_ax) = plt.subplots(1, 2, figsize=(16, 8))
for ax, title in ((gt_ax, "GT"), (pred_ax, "Prediction")):
    ax.imshow(image)
    ax.set_title(title)
    ax.axis("off")

# GT: YOLO 형식은 정규화된 (중심 x, 중심 y, 너비, 높이)
if label_path.exists():
    for line in label_path.read_text(encoding="utf-8").splitlines():
        class_id, cx, cy, bw, bh = map(float, line.split())
        x1 = (cx - bw / 2) * width
        y1 = (cy - bh / 2) * height
        box_width = bw * width
        box_height = bh * height

        gt_ax.add_patch(Rectangle(
            (x1, y1), box_width, box_height,
            fill=False, edgecolor="lime", linewidth=2
        ))
        gt_ax.text(x1, y1, trained_model.names[int(class_id)],
                   color="lime", fontsize=9)

# 예측: xyxy는 이미지 픽셀 좌표
for box in prediction.boxes:
    x1, y1, x2, y2 = box.xyxy[0].tolist()
    class_id = int(box.cls[0])
    confidence = float(box.conf[0])

    pred_ax.add_patch(Rectangle(
        (x1, y1), x2 - x1, y2 - y1,
        fill=False, edgecolor="red", linewidth=2
    ))
    pred_ax.text(x1, y1, f"{trained_model.names[class_id]} {confidence:.2f}",
                 color="red", fontsize=9)

plt.tight_layout()
plt.show()
```

```python
import random

image_paths = [
    p for p in (TACO_DIR / "images" / "val").iterdir()
    if p.suffix.lower() in {".jpg", ".jpeg", ".png"}
]
selected = random.Random(42).sample(image_paths, 10)

fig, axes = plt.subplots(10, 2, figsize=(14, 40))

for row, image_path in enumerate(selected):
    image = Image.open(image_path).convert("RGB")
    width, height = image.size
    label_path = TACO_DIR / "labels" / "val" / f"{image_path.stem}.txt"
    prediction = trained_model.predict(image, conf=0.10, verbose=False)[0]

    gt_ax, pred_ax = axes[row]
    for ax, title in ((gt_ax, "GT"), (pred_ax, "Prediction")):
        ax.imshow(image)
        ax.set_title(f"{image_path.name} — {title}")
        ax.axis("off")

    if label_path.exists():
        for line in label_path.read_text(encoding="utf-8").splitlines():
            class_id, cx, cy, bw, bh = map(float, line.split())
            x1 = (cx - bw / 2) * width
            y1 = (cy - bh / 2) * height

            gt_ax.add_patch(Rectangle(
                (x1, y1), bw * width, bh * height,
                fill=False, edgecolor="lime", linewidth=2
            ))
            gt_ax.text(x1, y1, trained_model.names[int(class_id)],
                       color="lime", fontsize=9)

    for box in prediction.boxes:
        x1, y1, x2, y2 = box.xyxy[0].tolist()
        class_id = int(box.cls[0])

        pred_ax.add_patch(Rectangle(
            (x1, y1), x2 - x1, y2 - y1,
            fill=False, edgecolor="red", linewidth=2
        ))
        pred_ax.text(
            x1, y1,
            f"{trained_model.names[class_id]} {float(box.conf[0]):.2f}",
            color="red", fontsize=9
        )

plt.tight_layout()
plt.show()
```

```python
from pathlib import Path
import csv
import yaml
import wandb

run_dir = Path("train")
run_dir.exists()
```

```python
wandb.login()
with wandb.init(
  project="taco-yolo",
  name="yolo11n-baseline",
  config=yaml.safe_load((run_dir / "args.yaml").read_text(encoding="utf-8")),
) as run:
  with (run_dir / "results.csv").open(newline="", encoding="utf-8") as file:
    for row in csv.DictReader(file):
      run.log({
        "epoch": int(row["epoch"]),
        **{key: float(value) for key, value in row.items() if key != "epoch"}
      })

  run.log({
    "validation/labels": wandb.Image(str(run_dir / "val_batch0_labels.jpg")),
    "validation/predictions": wandb.Image(str(run_dir / "val_batch0_pred.jpg"))
  })

  weights = wandb.Artifact("taco-yolo11n-baseline", type="model")
  weights.add_file(str(run_dir / "weights" / "best.pt"))
  weights.add_file(str(run_dir / "weights" / "last.pt"))
  run.log_artifact(weights)
```
